```python
import maxflow
import numpy as np
import cv2
import matplotlib.pyplot as plt

# 入力画像（グレースケール）
img = cv2.imread("./Image/cat.jpg", 0)
img = cv2.resize(img, (100, 100))  # 小さくして処理を軽く

rows, cols = img.shape
g = maxflow.Graph[float]()
nodes = g.add_grid_nodes(img.shape)

# 平滑化項（隣接ピクセル間の差が小さいほど同じラベルにしたい）
structure = np.array([[0, 1, 0],
                      [1, 0, 1],
                      [0, 1, 0]])
g.add_grid_edges(nodes, structure=structure, weights=50.0)

# データ項：明るさが一定値より高い→前景、低い→背景
foreground = img > 128
background = img <= 128

g.add_grid_tedges(nodes, foreground * 100, background * 100)

# 最小カットを実行
g.maxflow()
segments = g.get_grid_segments(nodes)

# マスク生成（True→背景、False→前景）
mask = np.logical_not(segments)

# 結果表示
plt.imshow(mask, cmap='gray')
plt.title("Graph Cut Segmentation")
plt.axis('off')
plt.show()
```
