# Data Load + Visualization
Load data in Python module, then visualize in this notebook.

```python
%matplotlib inline

import numpy as np
from study.test.src.dataload_pipeline.main import load_data
from study.test.src.dataload_pipeline.visualization import show_slice, overlay_multi_label, show_overlay
```

```python
result = load_data()

volume = result.volume_data.volume
volume_rotated = result.volume_rotated
label_map = result.label_map
contrast = result.contrast

print("volume shape:", volume.shape)
print("rotated shape:", volume_rotated.shape)
print("label shape:", label_map.shape)
print("contrast:", contrast)
```

```python
slice_index = 20
show_slice(volume, index=slice_index)
show_slice(volume_rotated, index=slice_index)
```

```python
labels = [
    label_map == 1,
    label_map == 2,
    label_map == 3,
    label_map == 4,
    label_map == 5,
]
colors = [
    [1.0, 0.9, 0.1],
    [0.5, 1.0, 0.2],
    [0.1, 0.9, 1.0],
    [0.5, 0.9, 0.5],
    [0.9, 0.5, 0.5],
]

volume_std = volume / contrast
overlay = overlay_multi_label(
    volume_std=volume_std,
    labels=labels,
    colors=colors,
    index=slice_index,
    threshold=0.0,
    border_only=False,
)
show_overlay(overlay, title=f"Slice {slice_index} with Labels")
```
